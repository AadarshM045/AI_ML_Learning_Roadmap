## Langchain Version V1

```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model=init_chat_model(model="openai/gpt-oss-120b",model_provider="groq")
responce=model.invoke("Tell me about programming languages?")
print(responce.content)
```

## LLM Model Integration with Google Gemini

```python
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "google_genai:gemini-3.8-flash",
    thinking_level="low"
)

response = model.invoke("Say hello in one sentence.")
print(response.text)
```
