# Pydantic Basics: Creating and Using Models

## Pydantic models are the foundation of data validation in Python whose Backend is in RUST. They use Python type annotations to define the structure and validate data at runtime. Here's a detailed exploration of basic model creation with several examples.

```python
from pydantic import BaseModel
from dataclasses import dataclass
```

```python
class Person(BaseModel):
    name:str
    age:int
    country:str

Person1=Person(name="Aadarsh",age=20,country="Nepal")
print(Person1)
```

```python
@dataclass
class Person():
    name:str
    age:int
    country:str

Person1=Person(name="Aadarsh",age=20,country= 34)
print(Person1)
```

### Why Do We Need Pydantic If We Already Have Dataclasses?

### 1. The Problem: Python Type Hints Don't Validate Data

### In Python, both dataclass and Pydantic use type annotations to describe the expected structure of data. However, describing the expected type is not the same as validating the actual value.

### For example:

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    country: str

person = Person(name="Aadarsh", age=20, country=34)

print(person)
```

## Why did this happen?

### Although country is annotated as str, Python's dataclass does not automatically enforce that type at runtime. It accepts the integer 34.

### This can become a problem when data comes from external sources, such as API requests, JSON files, users, or databases.

## 2. What Does Pydantic Do Differently?

### Pydantic is a Python library that uses type annotations to validate data at runtime and, when appropriate, convert values into the expected types.

### The same example using Pydantic:

```python
from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int
    country: str

person = Person(name="Aadarsh", age=20, country=34)

print(person)
```



## Output: a ValidationError, because the integer 34 is not accepted as a valid string by default in this case.

### Pydantic helps detect invalid data before it travels further into your application.

### Model With Optional Fields

### Add optional Fields using Pythons's Optional Type:
### Pydantic, Optional allows a field to accept either its specified data type or None. Setting a default value, such as = None, makes the field optional to omit. Pydantic also performs automatic type conversion (type casting) when possible, converting compatible input values to the expected type, such as "345" (string) to 345.0 (float).

```python
from typing import Optional

class Employee(BaseModel):
    emp_id:int
    name:str
    stream:str
    salary:Optional[float]=None
    is_active:Optional[bool]=True

emp1=Employee(emp_id=34,name="Jhon",stream="science")
print(emp1)

emp2=Employee(emp_id=56,name="The_Don",stream="Mob",salary=345532)
print(emp2)
```

### Nested

```python
from pydantic import BaseModel

class Address(BaseModel):
    city_name:str
    street:str
    zip_code:int

class info(BaseModel):
    name:str
    age:int
    address:Address

info1=info(
    name='Aadarsh',
    age=20,
    address={
        "city_name":"Lalitpur","street":"laxmi Marga","zip_code":"42232"
    }
)

print(info1)
```

## Pydantic Fields: Customization and Constraints

### The Field function in Pydantic enhances model fields beyond basic type hints by allowing you to specifc validation rules, default values, aliases, and more. Here's a comprehensive tutorial with examples.

```python
from pydantic import BaseModel,Field

class Student(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    age: int = Field(gt=0, le=120)

Student1=Student(name="Aadarsh",age=20)
print(Student1)
```

```python
from pydantic import BaseModel, Field

class Student(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    age: int = Field(gt=0, le=120)
    gpa: float = Field(default=0.0, ge=0.0, le=4.0)
    subjects: list[str] = Field(default_factory=list)

student = Student(
    name="Aadarsh",
    age="20",
    gpa=3.5,
    subjects=["Python", "Mathematics"]
)

print(student)
print(student.age)
```

### `description` in Pydantic Fields

The `description` parameter in `Field()` provides a human-readable explanation of a model field.

- It helps developers understand the purpose of each field.
- It is used for documentation and JSON Schema generation.
- It does not perform validation or change the field's value.

**Example:** `description="The age of the student in years"` explains what the `age` field represents.

**Note:** In FastAPI, these descriptions can appear in automatically generated API documentation, such as Swagger UI.

```python
from pydantic import BaseModel, Field

class Student(BaseModel):
    name: str = Field(
        min_length=2,
        max_length=50,
        description="The full name of the student"
    )

    age: int = Field(
        gt=0,
        le=120,
        description="The age of the student in years"
    )

student1 = Student(name="Aadarsh", age=20)

print(student1)
print(Student.model_json_schema())
```
