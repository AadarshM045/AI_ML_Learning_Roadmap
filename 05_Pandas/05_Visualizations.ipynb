## Visualizations

```python
import pandas as pd
```

```python
df=pd.read_csv("/home/aadarsh/Downloads/Ice Cream Ratings.csv")
df=df.set_index('Date')
df
```

```python
df.plot() ## Default is line
```

```python
df.plot(kind='line',subplots=True)
```

```python
df.plot(kind='line',title='IceCreams')
```

```python
df.plot(kind='line',title='IceCreams',xlabel = 'Daily ratings',ylabel = 'Scores')
```

```python
df.plot(kind='bar',stacked=True)
```

```python
df['Flavor Rating'].plot(kind='bar',stacked=True)
```

```python
df.plot.barh(stacked=True)
```

```python
df.plot.scatter(x='Flavor Rating',y='Overall Rating',s=50,c='Black')
```

```python
df.boxplot()
```

```python
df.plot.area()
```

```python
df.plot.pie(y='Flavor Rating',figsize=(10,5))
```
