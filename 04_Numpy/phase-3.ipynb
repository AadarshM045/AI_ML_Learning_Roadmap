### Advance Operations with Buisiness Examples


```python
import numpy as np
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani      
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites      
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub       
    [4, 180000, 210000, 240000, 270000],  # Burger Point     
    [5, 160000, 185000, 205000, 230000]   # Chai Point      
])
print("=======Zomato Sales Analysis=======")
print("Data Shape:-",sales_data.shape)
print("Data of First three: ",sales_data[:3])
```

```python
## Total Sales per Year
print("Per Year Sale:-",np.sum(sales_data,axis=0))
print("Restro Sale:-",np.sum(sales_data,axis=1))

## Minimum Sales
print(np.min(sales_data[:,1:],axis=1))

##Maximum sales per Year
print(np.max(sales_data[: ,1:],axis=0))

##Average Sales per Resturant 
print(np.mean(sales_data[: ,1:],axis=1))

##Cummulative Sales
print(np.cumsum)
 
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])

print("Vector Addition: ",vector1+vector2)
print("Product: ",vector1*vector2)
print("Dot Product: ",np.dot(vector1,vector2))
```

```python
resturant_types=np.array(["Paradise Biryani","Beijing Bites","Pizza Hub","Burger Point","Chai Point"])
vectorize_Operations=np.vectorize(str.upper)
print("In Uppercase: ",vectorize_Operations(resturant_types))
```

```python
##Monthly Average
monthly_avg=sales_data[: ,1:]/12
print("Monthly Average: ",monthly_avg)
```
