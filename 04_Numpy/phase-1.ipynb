```python
import numpy as np
zeros=np.zeros((3,4))
print(f"Zeros matrix:- {zeros}")
ones=np.ones((3,4))
print(f"Ones:- {ones}")
full=np.full((3,4),5)
print(f"Full:- {full}")
print(full*zeros)
random=np.random.random((3,4))
print(random)
sequence=np.arange(0,11,3)
print(sequence)
```

### Vector Tensor and Matrix

```python
import numpy as np
vector=np.array([3,5,7])
print(f"Vector:- {vector}")

matrix=np.array([[3,5,6],
                [6,3,6]])
print(f"Matrix:- {matrix}")

tensor=np.array([[[3,4],[5,6]],
                [[6,7],[7,3]]])
print(f"Tensor:- {tensor}")
```

## Array properties


```python
arr=np.array([[3,4],
                [5,7]])
print("Shape: ",arr.shape)
print("Dimensions: ",arr.ndim)
print("DType: ",arr.dtype)
print("Size: ",arr.size)
```

## Array Reshaping


```python
arr=np.array([[3,4,5],
                [4,7,2]])
print("reshaped: ",arr.reshape(3,2))
print("Array: ",arr)
print("Flattened array: ",arr.flatten()) #changes main array
print("Array: ",arr.ravel())#return view instead of copy
#Transpose
arr=np.array([[3,4,5],
                [4,7,2]])
print("Transpose: ",arr.T)
```
